# Pull × Air Quadrant Analysis
Builds parallel binary classifiers for `is_pull` and `is_air` to
identify where MLB hitters fall in the Pull × Air quadrant system.

**Quadrants** (split on median Pull% and median Air Ball%):
- **Both (Pull + Air)** — high pull, high air
- **Pull only** — high pull, low air
- **Air only** — low pull, high air
- **Neither** — low pull, low air

**Approach**
- Parallel binary classifiers for Pull and Air (not cascade — avoids the same-angle problem)
- **Explanatory:** Logistic Regression with interpretable coefficients
- **Predictive:** Random Forest to capture angle×tilt interactions
- **Subset models** for Both vs Neither (Tilt as tiebreaker) and Pull only vs Air only (angles as discriminator)
- 2D probability scatter to show how quadrants live in (P(pull), P(air)) space

**Sections**
1. Imports & config
2. Load data, build features, assign quadrants
3. Feature distributions by quadrant
4. Parallel binary classifiers — Logistic Regression (explanatory)
5. Parallel binary classifiers — Random Forest (predictive)
6. Confusion matrices & calibration
7. SHAP feature importance
8. Subset model — Both vs Neither (Tilt isolation)
9. Subset model — Pull only vs Air only (Angles isolation)
10. 2D probability scatter


## 1. Imports & Config

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import shap
from pathlib import Path
from scipy.stats import spearmanr, pearsonr

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import (
    classification_report, confusion_matrix, roc_auc_score,
    brier_score_loss, RocCurveDisplay, PrecisionRecallDisplay
)
from sklearn.calibration import calibration_curve, CalibratedClassifierCV

pd.set_option("display.float_format", "{:.4f}".format)

SEED         = 42
MIN_BBE      = 300
CV_FOLDS     = 5

# ── Feature set
FEATURES = [
    "mean_tilt",         # Swing Path Tilt
    "mean_aa",           # Attack Angle
    "mean_dir",          # Attack Direction (Statcast convention)
    "mean_bs",           # Mean Bat Speed
    "bs_p90",            # Peak bat speed
    "hard_swing_rate",
    "z_swing_pct",       # In-zone swing rate
    "o_swing_pct",       # Out-of-zone swing rate (chase)
    "fb_swing_share",    # Fastball share of swings
]

QUAD_COLORS = {
    "Neither":           "lightgray",
    "Pull only":         "#c44e52",
    "Air only":          "steelblue",
    "Both (Pull + Air)": "green",
}
QUAD_ORDER = ["Neither", "Pull only", "Air only", "Both (Pull + Air)"]

print("Imports OK")
```

## 2. Load Data, Build Features, Assign Quadrants

```python
# Assumes batter_df already exists from prior notebook OR load from cache
# Required columns: pull_pct, air_ball_pct, plus all FEATURES, plus 'batter'

# Load batter_df from cache (replace path if needed)
try:
    batter_df = pd.read_parquet("batter_df.parquet")
    print(f"Loaded batter_df from cache: {batter_df.shape}")
except FileNotFoundError:
    print("batter_df.parquet not found — paste your batter-level DataFrame here.")
    print("Expected columns:", ["pull_pct","air_ball_pct","n_bbe"] + FEATURES + ["batter"])

# ── Apply BBE filter
print(f"\nBefore BBE filter: {len(batter_df)} rows")
batter_df = batter_df[batter_df["n_bbe"] >= MIN_BBE].reset_index(drop=True)
print(f"After  BBE filter (≥{MIN_BBE}): {len(batter_df)} rows")

# ── Quadrant assignment on medians
pull_cut = batter_df["pull_pct"].median()
air_cut  = batter_df["air_ball_pct"].median()

batter_df["is_pull"] = (batter_df["pull_pct"]     >= pull_cut).astype(int)
batter_df["is_air"]  = (batter_df["air_ball_pct"] >= air_cut ).astype(int)

def label_quadrant(row):
    if row["is_pull"] and row["is_air"]:      return "Both (Pull + Air)"
    if row["is_pull"] and not row["is_air"]:  return "Pull only"
    if not row["is_pull"] and row["is_air"]:  return "Air only"
    return "Neither"

batter_df["quadrant"] = batter_df.apply(label_quadrant, axis=1)

print(f"\nMedian cutoffs:  pull_pct={pull_cut:.3f}  air_ball_pct={air_cut:.3f}")
print(f"\nQuadrant sizes:")
print(batter_df["quadrant"].value_counts().reindex(QUAD_ORDER))

# Drop rows with missing features
missing = batter_df[FEATURES].isna().any(axis=1).sum()
print(f"\nRows with missing features: {missing}  (will be dropped from models)")
```

## 3. Feature Distributions by Quadrant

```python
# Replicate the diagnostic bar plots from earlier — verify the Both/Neither story
fig, axes = plt.subplots(2, 3, figsize=(15, 9))
axes = axes.flatten()

key_feats = ["mean_tilt", "mean_aa", "mean_dir",
              "mean_bs", "z_swing_pct", "o_swing_pct"]
labels    = ["Swing Path Tilt", "Attack Angle", "Attack Direction",
              "Bat Speed", "Z-Swing %", "O-Swing %"]

for ax, feat, lab in zip(axes, key_feats, labels):
    means = batter_df.groupby("quadrant")[feat].mean().reindex(QUAD_ORDER)
    bars  = ax.bar(QUAD_ORDER, means.values,
                    color=[QUAD_COLORS[q] for q in QUAD_ORDER],
                    edgecolor="white")
    for bar, val in zip(bars, means.values):
        ax.text(bar.get_x() + bar.get_width()/2, val,
                f"{val:.3f}", ha="center", va="bottom", fontsize=9)
    ax.set_title(lab)
    ax.set_xticklabels(QUAD_ORDER, rotation=20, ha="right", fontsize=8)
    ax.grid(axis="y", alpha=0.3)

plt.suptitle("Feature Means by Quadrant", y=1.01, fontsize=12)
plt.tight_layout()
plt.show()
```

## 4. Parallel Binary Classifiers — Logistic Regression (Explanatory)
Two independent binary models: `is_pull` and `is_air`. Coefficients tell the
direct story of which features push toward pulling and which push toward elevating.


```python
sub      = batter_df[FEATURES + ["is_pull", "is_air", "quadrant", "pull_pct", "air_ball_pct", "batter"]].dropna()
X        = sub[FEATURES].values
y_pull   = sub["is_pull"].values
y_air    = sub["is_air"].values
quadrant = sub["quadrant"].values

print(f"Model dataset: n={len(sub)}")
print(f"  is_pull: {y_pull.sum()} positive, {len(y_pull) - y_pull.sum()} negative")
print(f"  is_air:  {y_air.sum()} positive,  {len(y_air)  - y_air.sum()} negative")

def fit_lr(X, y, label):
    pipe = Pipeline([
        ("scaler", StandardScaler()),
        ("lr",     LogisticRegression(
            solver="lbfgs", class_weight="balanced",
            max_iter=1000, random_state=SEED, C=0.5,
        )),
    ])
    cv         = StratifiedKFold(n_splits=CV_FOLDS, shuffle=True, random_state=SEED)
    y_pred     = cross_val_predict(pipe, X, y, cv=cv)
    y_proba    = cross_val_predict(pipe, X, y, cv=cv, method="predict_proba")[:, 1]
    auc        = roc_auc_score(y, y_proba)
    brier      = brier_score_loss(y, y_proba)
    print(f"\n── {label} (Logistic Regression) ─────────────")
    print(f"   AUC : {auc:.4f}   Brier: {brier:.4f}")
    print(classification_report(y, y_pred, target_names=[f"Not {label}", label],
                                zero_division=0))
    # Fit on full data for coefficient extraction
    pipe.fit(X, y)
    return pipe, y_pred, y_proba, auc, brier

lr_pull, lr_pull_pred, lr_pull_proba, lr_pull_auc, lr_pull_brier =     fit_lr(X, y_pull, "is_pull")
lr_air,  lr_air_pred,  lr_air_proba,  lr_air_auc,  lr_air_brier  =     fit_lr(X, y_air,  "is_air")
```

```python
# ── Coefficient comparison ─────────────────────────────────────────────────
coefs = pd.DataFrame({
    "Feature":    FEATURES,
    "is_pull β":  lr_pull.named_steps["lr"].coef_[0],
    "is_air β":   lr_air.named_steps["lr"].coef_[0],
})
coefs["|pull|+|air|"] = coefs["is_pull β"].abs() + coefs["is_air β"].abs()
coefs = coefs.sort_values("|pull|+|air|", ascending=False).drop(columns=["|pull|+|air|"])

print("Standardised Coefficients (sorted by combined magnitude)")
print(coefs.to_string(index=False))

# ── Plot
fig, axes = plt.subplots(1, 2, figsize=(13, max(3, len(FEATURES) * 0.4)))
for ax, col, label in [
    (axes[0], "is_pull β", "is_pull"),
    (axes[1], "is_air β",  "is_air"),
]:
    sorted_c = coefs.sort_values(col, key=lambda x: x.abs(), ascending=True)
    colors   = ["green" if v > 0 else "red" for v in sorted_c[col]]
    ax.barh(sorted_c["Feature"], sorted_c[col], color=colors, alpha=0.8)
    ax.axvline(0, color="black", linewidth=0.6)
    ax.set_xlabel("Standardised coefficient")
    ax.set_title(f"{label}  (AUC={lr_pull_auc if label=='is_pull' else lr_air_auc:.3f})")

plt.suptitle("Logistic Regression — Feature Coefficients", y=1.02, fontsize=12)
plt.tight_layout()
plt.show()
```

## 5. Parallel Binary Classifiers — Random Forest (Predictive)
Trees capture interaction effects that linear models miss — especially the
non-monotonic angle × tilt relationship that distinguishes Both from Neither.


```python
def fit_rf(X, y, label):
    pipe = Pipeline([
        ("scaler", StandardScaler()),
        ("rf",     RandomForestClassifier(
            n_estimators=400, max_depth=8, min_samples_leaf=10,
            class_weight="balanced", random_state=SEED, n_jobs=-1,
        )),
    ])
    cv      = StratifiedKFold(n_splits=CV_FOLDS, shuffle=True, random_state=SEED)
    y_pred  = cross_val_predict(pipe, X, y, cv=cv)
    y_proba = cross_val_predict(pipe, X, y, cv=cv, method="predict_proba")[:, 1]
    auc     = roc_auc_score(y, y_proba)
    brier   = brier_score_loss(y, y_proba)
    print(f"\n── {label} (Random Forest) ──────────────────")
    print(f"   AUC : {auc:.4f}   Brier: {brier:.4f}")
    print(classification_report(y, y_pred, target_names=[f"Not {label}", label],
                                zero_division=0))
    pipe.fit(X, y)
    return pipe, y_pred, y_proba, auc, brier

rf_pull, rf_pull_pred, rf_pull_proba, rf_pull_auc, rf_pull_brier =     fit_rf(X, y_pull, "is_pull")
rf_air,  rf_air_pred,  rf_air_proba,  rf_air_auc,  rf_air_brier  =     fit_rf(X, y_air,  "is_air")
```

```python
# ── ROC + PR curves: LR vs RF
fig, axes = plt.subplots(2, 2, figsize=(13, 9))

for col, (y_true, lr_p, rf_p, lr_auc, rf_auc, title) in enumerate([
    (y_pull, lr_pull_proba, rf_pull_proba, lr_pull_auc, rf_pull_auc, "is_pull"),
    (y_air,  lr_air_proba,  rf_air_proba,  lr_air_auc,  rf_air_auc,  "is_air"),
]):
    RocCurveDisplay.from_predictions(y_true, lr_p, name="Logistic Regression",
                                       ax=axes[0, col], color="#1565C0")
    RocCurveDisplay.from_predictions(y_true, rf_p, name="Random Forest",
                                       ax=axes[0, col], color="#2E7D32")
    axes[0, col].plot([0,1],[0,1],"k--",linewidth=0.6)
    axes[0, col].set_title(f"ROC — {title}")

    PrecisionRecallDisplay.from_predictions(y_true, lr_p, name="Logistic Regression",
                                              ax=axes[1, col], color="#1565C0")
    PrecisionRecallDisplay.from_predictions(y_true, rf_p, name="Random Forest",
                                              ax=axes[1, col], color="#2E7D32")
    axes[1, col].axhline(y_true.mean(), color="gray", linestyle="--",
                          linewidth=0.6, label=f"Baseline ({y_true.mean():.3f})")
    axes[1, col].set_title(f"Precision-Recall — {title}")
    axes[1, col].legend(fontsize=8)

plt.suptitle("Logistic Regression vs Random Forest", y=1.01, fontsize=12)
plt.tight_layout()
plt.show()
```

## 6. Confusion Matrices & Calibration

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 9))

for row, (model_name, pull_pred, air_pred) in enumerate([
    ("Logistic Regression", lr_pull_pred, lr_air_pred),
    ("Random Forest",        rf_pull_pred, rf_air_pred),
]):
    for col, (y_true, y_pred, label) in enumerate([
        (y_pull, pull_pred, "is_pull"),
        (y_air,  air_pred,  "is_air"),
    ]):
        cm = confusion_matrix(y_true, y_pred)
        sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                     xticklabels=[f"Not {label}", label],
                     yticklabels=[f"Not {label}", label],
                     ax=axes[row, col], cbar=False)
        axes[row, col].set_xlabel("Predicted")
        axes[row, col].set_ylabel("Actual")
        axes[row, col].set_title(f"{model_name} — {label}")

plt.suptitle("Confusion Matrices", y=1.01, fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# ── Calibration curves
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

for ax, (y_true, lr_p, rf_p, title) in zip(axes, [
    (y_pull, lr_pull_proba, rf_pull_proba, "is_pull"),
    (y_air,  lr_air_proba,  rf_air_proba,  "is_air"),
]):
    for probs, label, color in [
        (lr_p, "Logistic Regression", "#1565C0"),
        (rf_p, "Random Forest",       "#2E7D32"),
    ]:
        frac, mean_p = calibration_curve(y_true, probs, n_bins=10, strategy="quantile")
        ax.plot(mean_p, frac, marker="o", markersize=5,
                color=color, linewidth=1.5, label=label)
    ax.plot([0,1],[0,1],"k--",linewidth=0.6,label="Perfect calibration")
    ax.set_xlabel("Mean predicted probability")
    ax.set_ylabel("Fraction of positives")
    ax.set_title(f"Calibration — {title}")
    ax.legend(fontsize=9)

plt.suptitle("Calibration Curves", y=1.02, fontsize=12)
plt.tight_layout()
plt.show()

print("Brier scores:")
print(f"  is_pull   LR={lr_pull_brier:.4f}  RF={rf_pull_brier:.4f}")
print(f"  is_air    LR={lr_air_brier:.4f}  RF={rf_air_brier:.4f}")
```

## 7. SHAP Feature Importance (Random Forest)

```python
# Refit RF directly (without pipeline) on standardised data for SHAP
scaler  = StandardScaler().fit(X)
X_std   = scaler.transform(X)

rf_p_raw = RandomForestClassifier(
    n_estimators=400, max_depth=8, min_samples_leaf=10,
    class_weight="balanced", random_state=SEED, n_jobs=-1,
).fit(X_std, y_pull)

rf_a_raw = RandomForestClassifier(
    n_estimators=400, max_depth=8, min_samples_leaf=10,
    class_weight="balanced", random_state=SEED, n_jobs=-1,
).fit(X_std, y_air)

# SHAP
exp_p = shap.TreeExplainer(rf_p_raw)
exp_a = shap.TreeExplainer(rf_a_raw)

shap_p = exp_p.shap_values(X_std)
shap_a = exp_a.shap_values(X_std)

# Handle both old and new SHAP API outputs
def get_positive_shap(shap_vals):
    if isinstance(shap_vals, list):
        return shap_vals[1]
    if shap_vals.ndim == 3:
        return shap_vals[:, :, 1]
    return shap_vals

shap_p_pos = get_positive_shap(shap_p)
shap_a_pos = get_positive_shap(shap_a)

print("SHAP computed.")
print(f"  is_pull SHAP shape: {shap_p_pos.shape}")
print(f"  is_air  SHAP shape: {shap_a_pos.shape}")
```

```python
# ── SHAP summary bar plots — mean |SHAP| by feature
def summary_bar(shap_vals, X_std, features, title, ax):
    mean_abs = np.abs(shap_vals).mean(axis=0)
    order    = np.argsort(mean_abs)
    ax.barh([features[i] for i in order],
            mean_abs[order], color="#2E7D32", alpha=0.85)
    ax.set_xlabel("mean |SHAP value|")
    ax.set_title(title)

fig, axes = plt.subplots(1, 2, figsize=(13, max(3, len(FEATURES) * 0.4)))
summary_bar(shap_p_pos, X_std, FEATURES, "is_pull — SHAP Importance", axes[0])
summary_bar(shap_a_pos, X_std, FEATURES, "is_air  — SHAP Importance", axes[1])
plt.suptitle("SHAP Feature Importance (Random Forest)", y=1.02, fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# ── SHAP beeswarm — show how each feature pushes predictions per hitter
fig, axes = plt.subplots(1, 2, figsize=(14, max(4, len(FEATURES) * 0.4)))

plt.sca(axes[0])
shap.summary_plot(shap_p_pos, pd.DataFrame(X_std, columns=FEATURES),
                  show=False, plot_type="dot")
axes[0].set_title("is_pull — SHAP per Hitter")

plt.sca(axes[1])
shap.summary_plot(shap_a_pos, pd.DataFrame(X_std, columns=FEATURES),
                  show=False, plot_type="dot")
axes[1].set_title("is_air — SHAP per Hitter")

plt.tight_layout()
plt.show()
```

## 8. Subset Model — Both vs Neither
The angles look identical between these groups. This isolates which features
(especially Tilt) actually distinguish them.


```python
# Filter to just Both + Neither
sub_bn      = sub[sub["quadrant"].isin(["Both (Pull + Air)", "Neither"])].copy()
sub_bn["y"] = (sub_bn["quadrant"] == "Both (Pull + Air)").astype(int)
X_bn        = sub_bn[FEATURES].values
y_bn        = sub_bn["y"].values

print(f"Both vs Neither — n={len(sub_bn)}  "
      f"(Both={y_bn.sum()}, Neither={len(y_bn) - y_bn.sum()})")

# Logistic Regression
pipe_bn_lr = Pipeline([
    ("scaler", StandardScaler()),
    ("lr",     LogisticRegression(solver="lbfgs", class_weight="balanced",
                                    max_iter=1000, random_state=SEED, C=0.5)),
])
cv             = StratifiedKFold(n_splits=CV_FOLDS, shuffle=True, random_state=SEED)
y_bn_pred_lr   = cross_val_predict(pipe_bn_lr, X_bn, y_bn, cv=cv)
y_bn_proba_lr  = cross_val_predict(pipe_bn_lr, X_bn, y_bn, cv=cv, method="predict_proba")[:,1]
auc_bn_lr      = roc_auc_score(y_bn, y_bn_proba_lr)
pipe_bn_lr.fit(X_bn, y_bn)

# Random Forest
pipe_bn_rf = Pipeline([
    ("scaler", StandardScaler()),
    ("rf",     RandomForestClassifier(
        n_estimators=400, max_depth=8, min_samples_leaf=10,
        class_weight="balanced", random_state=SEED, n_jobs=-1)),
])
y_bn_pred_rf   = cross_val_predict(pipe_bn_rf, X_bn, y_bn, cv=cv)
y_bn_proba_rf  = cross_val_predict(pipe_bn_rf, X_bn, y_bn, cv=cv, method="predict_proba")[:,1]
auc_bn_rf      = roc_auc_score(y_bn, y_bn_proba_rf)
pipe_bn_rf.fit(X_bn, y_bn)

print(f"\nLR AUC: {auc_bn_lr:.4f}   RF AUC: {auc_bn_rf:.4f}")

# LR coefficients
coefs_bn = pd.DataFrame({
    "Feature": FEATURES,
    "Coef":    pipe_bn_lr.named_steps["lr"].coef_[0],
}).sort_values("Coef", key=lambda x: x.abs(), ascending=False)

print("\nLR Coefficients (Both = positive):")
print(coefs_bn.to_string(index=False))
```

```python
# ── Plot: LR coefficients + confusion matrices for Both vs Neither
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# LR coefs
sorted_c = coefs_bn.sort_values("Coef", key=lambda x: x.abs(), ascending=True)
colors   = ["green" if v > 0 else "red" for v in sorted_c["Coef"]]
axes[0].barh(sorted_c["Feature"], sorted_c["Coef"], color=colors, alpha=0.8)
axes[0].axvline(0, color="black", linewidth=0.6)
axes[0].set_xlabel("Standardised coefficient")
axes[0].set_title(f"Both vs Neither — LR coefs\n(AUC={auc_bn_lr:.3f})")

# CMs
for ax, y_pred, model_label, auc in [
    (axes[1], y_bn_pred_lr, "Logistic Regression", auc_bn_lr),
    (axes[2], y_bn_pred_rf, "Random Forest",        auc_bn_rf),
]:
    cm = confusion_matrix(y_bn, y_pred)
    sns.heatmap(cm, annot=True, fmt="d", cmap="Greens",
                xticklabels=["Neither","Both"], yticklabels=["Neither","Both"],
                ax=ax, cbar=False)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")
    ax.set_title(f"{model_label}\n(AUC={auc:.3f})")

plt.tight_layout()
plt.show()
```

## 9. Subset Model — Pull only vs Air only
These two have opposite angles but neither both. This isolates how angles
discriminate when tilt is roughly comparable.


```python
sub_pa      = sub[sub["quadrant"].isin(["Pull only", "Air only"])].copy()
sub_pa["y"] = (sub_pa["quadrant"] == "Pull only").astype(int)
X_pa        = sub_pa[FEATURES].values
y_pa        = sub_pa["y"].values

print(f"Pull only vs Air only — n={len(sub_pa)}  "
      f"(Pull only={y_pa.sum()}, Air only={len(y_pa) - y_pa.sum()})")

pipe_pa_lr = Pipeline([
    ("scaler", StandardScaler()),
    ("lr",     LogisticRegression(solver="lbfgs", class_weight="balanced",
                                    max_iter=1000, random_state=SEED, C=0.5)),
])
y_pa_pred_lr   = cross_val_predict(pipe_pa_lr, X_pa, y_pa, cv=cv)
y_pa_proba_lr  = cross_val_predict(pipe_pa_lr, X_pa, y_pa, cv=cv, method="predict_proba")[:,1]
auc_pa_lr      = roc_auc_score(y_pa, y_pa_proba_lr)
pipe_pa_lr.fit(X_pa, y_pa)

pipe_pa_rf = Pipeline([
    ("scaler", StandardScaler()),
    ("rf",     RandomForestClassifier(
        n_estimators=400, max_depth=8, min_samples_leaf=10,
        class_weight="balanced", random_state=SEED, n_jobs=-1)),
])
y_pa_pred_rf   = cross_val_predict(pipe_pa_rf, X_pa, y_pa, cv=cv)
y_pa_proba_rf  = cross_val_predict(pipe_pa_rf, X_pa, y_pa, cv=cv, method="predict_proba")[:,1]
auc_pa_rf      = roc_auc_score(y_pa, y_pa_proba_rf)
pipe_pa_rf.fit(X_pa, y_pa)

print(f"\nLR AUC: {auc_pa_lr:.4f}   RF AUC: {auc_pa_rf:.4f}")

coefs_pa = pd.DataFrame({
    "Feature": FEATURES,
    "Coef":    pipe_pa_lr.named_steps["lr"].coef_[0],
}).sort_values("Coef", key=lambda x: x.abs(), ascending=False)

print("\nLR Coefficients (Pull only = positive):")
print(coefs_pa.to_string(index=False))

# Plot
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sorted_c = coefs_pa.sort_values("Coef", key=lambda x: x.abs(), ascending=True)
colors   = ["green" if v > 0 else "red" for v in sorted_c["Coef"]]
axes[0].barh(sorted_c["Feature"], sorted_c["Coef"], color=colors, alpha=0.8)
axes[0].axvline(0, color="black", linewidth=0.6)
axes[0].set_xlabel("Standardised coefficient")
axes[0].set_title(f"Pull only vs Air only — LR coefs\n(AUC={auc_pa_lr:.3f})")

for ax, y_pred, model_label, auc in [
    (axes[1], y_pa_pred_lr, "Logistic Regression", auc_pa_lr),
    (axes[2], y_pa_pred_rf, "Random Forest",        auc_pa_rf),
]:
    cm = confusion_matrix(y_pa, y_pred)
    sns.heatmap(cm, annot=True, fmt="d", cmap="Oranges",
                xticklabels=["Air only","Pull only"],
                yticklabels=["Air only","Pull only"],
                ax=ax, cbar=False)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")
    ax.set_title(f"{model_label}\n(AUC={auc:.3f})")

plt.tight_layout()
plt.show()
```

## 10. 2D Probability Scatter
Predicted (P(pull), P(air)) for every hitter, coloured by true quadrant.
Shows how cleanly the parallel models separate the quadrants in probability space.


```python
fig, axes = plt.subplots(1, 2, figsize=(15, 7))

for ax, (pull_proba, air_proba, model_label) in zip(axes, [
    (lr_pull_proba, lr_air_proba, "Logistic Regression"),
    (rf_pull_proba, rf_air_proba, "Random Forest"),
]):
    for quad in QUAD_ORDER:
        mask = quadrant == quad
        ax.scatter(pull_proba[mask], air_proba[mask],
                   color=QUAD_COLORS[quad], alpha=0.6, s=20,
                   label=f"{quad}  (n={mask.sum()})")

    ax.axhline(0.5, color="black", linestyle=":", linewidth=0.8)
    ax.axvline(0.5, color="black", linestyle=":", linewidth=0.8)

    # Annotate quadrants
    ax.text(0.05, 0.95, "Air only",  fontsize=10, color="steelblue",
            transform=ax.transAxes)
    ax.text(0.75, 0.95, "Both",      fontsize=10, color="green",
            transform=ax.transAxes)
    ax.text(0.05, 0.03, "Neither",   fontsize=10, color="gray",
            transform=ax.transAxes)
    ax.text(0.75, 0.03, "Pull only", fontsize=10, color="#c44e52",
            transform=ax.transAxes)

    ax.set_xlabel("Predicted P(is_pull)")
    ax.set_ylabel("Predicted P(is_air)")
    ax.set_title(f"{model_label}\n2D Probability Scatter")
    ax.legend(fontsize=8, loc="center right", bbox_to_anchor=(1, 0.5))
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)

plt.suptitle("Pull × Air Quadrant Predictions in Probability Space", y=1.01, fontsize=13)
plt.tight_layout()
plt.show()
```

```python
# ── 4-quadrant confusion matrix: predicted quadrant from joint probabilities
def assign_quadrant(p_pull, p_air, t_pull=0.5, t_air=0.5):
    if p_pull >= t_pull and p_air >= t_air:    return "Both (Pull + Air)"
    if p_pull >= t_pull and p_air <  t_air:    return "Pull only"
    if p_pull <  t_pull and p_air >= t_air:    return "Air only"
    return "Neither"

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

for ax, (pull_proba, air_proba, model_label) in zip(axes, [
    (lr_pull_proba, lr_air_proba, "Logistic Regression"),
    (rf_pull_proba, rf_air_proba, "Random Forest"),
]):
    pred_quads = np.array([assign_quadrant(p, a)
                            for p, a in zip(pull_proba, air_proba)])
    cm         = pd.crosstab(
        pd.Series(quadrant,   name="Actual"),
        pd.Series(pred_quads, name="Predicted"),
    ).reindex(index=QUAD_ORDER, columns=QUAD_ORDER, fill_value=0)
    cm_norm    = cm.div(cm.sum(axis=1), axis=0)

    sns.heatmap(cm_norm, annot=cm, fmt="d", cmap="Blues",
                linewidths=0.5, linecolor="white", ax=ax, cbar=False)
    ax.set_title(f"{model_label}\nQuadrant Confusion (row-normalised)")
    ax.set_xlabel("Predicted Quadrant")
    ax.set_ylabel("Actual Quadrant")

plt.tight_layout()
plt.show()
```

## Summary

**Predictive (Random Forest):** Use the (P(pull), P(air)) probabilities for projection.
Hitters in the corners are high-confidence quadrant assignments; hitters near (0.5, 0.5)
are genuinely ambiguous and worth scouting attention.

**Explanatory (Logistic Regression):** Use the coefficients for development conversations.

**Both vs Neither:** The angles don't separate these — Tilt is the discriminator.
**Pull only vs Air only:** Tilt is roughly equivalent — angles are the discriminators.

The 2D probability scatter is the deliverable for scouting. Hitters with high P(both)
in the predictive model are projection candidates; hitters with split probabilities
(high pull, low air) reveal the specific developmental gap.
